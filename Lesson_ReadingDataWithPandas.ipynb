# Lesson: Reading in Data With Pandas

To read in a csv file, we will use the pandas function <b> read_csv(). </b>

<b> pd.read_csv(name_of_file) </b>

This will read our csv file into a pandas DataFrame. You can think of a <b> DataFrame </b> as a table of data with rows and columns. 

Here is an example where we have read in a file called <b>study.csv </b>

```python
import pandas as pd

data = pd.read_csv("study.csv")
print(data)
```

You’ll have noticed that there are two columns in study.csv, these are:

-  <b> Time Spent Studying (hours) </b>

-  <b> Exam Mark (%) </b>

We can extract these out into <b> x </b> our independent variable and <b> y </b> our dependent variable. 

<b> x </b> and <b> y </b> are pandas <b> series </b> objects. 

These are like 1-dimensional arrays.

To convert these to numpy arrays we use:

-  <b> series.to_numpy() </b>

```python
import pandas as pd

data = pd.read_csv("study.csv")

# Remove extra spaces from column headings
data.columns = data.columns.str.strip()

x = data["Time Spent Studying (hours)"].to_numpy()
y = data["Exam Mark (%)"].to_numpy()

print(x)
print(y)
```

We use numpy arrays because it’s easier to manipulate their dimensions using <b> .reshape(rows, columns) </b>

This allows you to quickly change your 1D vector into a 2D column or row vector. You’ll see that this will be useful later.

```python
import pandas as pd

data = pd.read_csv("study.csv")

data.columns = data.columns.str.strip()

x = data["Time Spent Studying (hours)"].to_numpy()
y = data["Exam Mark (%)"].to_numpy()

# Reshape the arrays into columns
x = x.reshape(7, 1)
y = y.reshape(7, 1)

print(x)
print(y)
```

```python
import numpy as np

xy = np.column_stack((x, y))

print(xy)
```
